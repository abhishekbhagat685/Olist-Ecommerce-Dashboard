# Data Cleaning and Exploration

This notebook performs data quality checks on the Olist orders dataset.

```python
import zipfile

with zipfile.ZipFile("archive.zip", "r") as zip_ref:
    zip_ref.extractall("data")
    
```

```python
import os

os.listdir("data")
```

```python
import pandas as pd

orders = pd.read_csv("data/olist_orders_dataset.csv")
```

```python
orders.info()
```

```python
orders.isnull().sum()
```

```python
orders.describe()
```

```python
print("Rows:", orders.shape[0])
print("Columns:", orders.shape[1])
```

```python
missing = orders.isnull().sum()

missing_percentage = (
    orders.isnull().sum() / len(orders)
) * 100

pd.DataFrame({
    "Missing Values": missing,
    "Percentage": missing_percentage
}).sort_values(
    by="Missing Values",
    ascending=False
)
```

```python
orders.duplicated().sum()
```

```python
orders["order_id"].duplicated().sum()
```

```python
orders["order_status"].value_counts()
```

```python
orders["order_status"].value_counts().plot(
    kind="bar"
)
```

```python
date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_cols:
    orders[col] = pd.to_datetime(
        orders[col]
    )
```

```python
orders["order_purchase_timestamp"].min()
```

```python
print(
    "Data from",
    orders["order_purchase_timestamp"].min(),
    "to",
    orders["order_purchase_timestamp"].max()
)
```

```python
orders["year_month"] = (
    orders["order_purchase_timestamp"]
    .dt.to_period("M")
)
```

```python
monthly_orders = (
    orders.groupby("year_month")
    .size()
)
```

```python
monthly_orders.plot(
    figsize=(12,5)
)
```
