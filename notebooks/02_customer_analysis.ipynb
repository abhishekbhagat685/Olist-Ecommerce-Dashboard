```python
import matplotlib.pyplot as plt

customers["customer_state"]\
    .value_counts()\
    .head(10)\
    .plot(kind="bar")

plt.title("Top 10 Customer States")
plt.show()
```

```python
customers["customer_city"]\
    .value_counts()\
    .head(10)\
    .plot(kind="bar")

plt.title("Top 10 Customer Cities")
plt.show()
```

```python
orders = pd.read_csv(
    "data/olist_orders_dataset.csv"
)
```

```python
orders.shape
```

```python
customer_orders = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="inner"
)
```

```python
customer_orders.shape
```

```python
customer_orders.head()
```

```python
customer_orders.groupby(
    "customer_state"
)["order_id"].count()\
.sort_values(ascending=False)\
.head(10)
```

```python
customer_orders.groupby(
    "customer_state"
)["order_id"].count()\
.sort_values(ascending=False)\
.head(10)\
.plot(kind="bar")

plt.title("Orders by State")
plt.show()
```

Key Findings:

São Paulo contributes the highest number of customers.
Orders are concentrated in a few major Brazilian states.
Customer distribution is highly uneven across regions.
Customer and order datasets were successfully joined using customer_id.
