```python
import pandas as pd
import matplotlib.pyplot as plt

orders = pd.read_csv("data/olist_orders_dataset.csv")
items = pd.read_csv("data/olist_order_items_dataset.csv")
payments = pd.read_csv("data/olist_order_payments_dataset.csv")
products = pd.read_csv("data/olist_products_dataset.csv")
```

```python
print("Orders:", orders.shape)
print("Items:", items.shape)
print("Payments:", payments.shape)
print("Products:", products.shape)
```

```python
total_revenue = payments["payment_value"].sum()

print(f"Total Revenue: ${total_revenue:,.2f}")
```

```python
avg_order_value = payments["payment_value"].mean()

print(f"Average Order Value: ${avg_order_value:.2f}")
```

```python
total_orders = orders["order_id"].nunique()

print("Total Orders:", total_orders)
```

```python
revenue_per_order = total_revenue / total_orders

print(f"Revenue Per Order: ${revenue_per_order:.2f}")
```

```python
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)
```

```python
rders["month"] = orders[
    "order_purchase_timestamp"
].dt.to_period("M")
```

```python
monthly_orders = orders.groupby("month")["order_id"].count()

monthly_orders.head()
```

```python
monthly_orders.plot(figsize=(12,5))

plt.title("Monthly Orders Trend")
plt.xlabel("Month")
plt.ylabel("Orders")
plt.show()
```
